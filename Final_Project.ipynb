###  Data Type Inspection and Conversion

In this step, we inspect the values in columns stored as strings and convert them to appropriate data types where necessary, ensuring the dataset is suitable for accurate analysis.

```python
# Inspect values before converting data types
df[['Year', 'mkt_fare', 'carpax', 'caravgfare']].head(10)
```

```python
# Remove formatting characters and convert columns to numeric types
df['Year'] = pd.to_numeric(df['Year'].str.replace(',', '', regex=False))
df['mkt_fare'] = pd.to_numeric(df['mkt_fare'].str.replace('$', '', regex=False).str.replace(',', '', regex=False))
df['carpax'] = pd.to_numeric(df['carpax'].str.replace(',', '', regex=False))
df['caravgfare'] = pd.to_numeric(df['caravgfare'].str.replace('$', '', regex=False).str.replace(',', '', regex=False))

# Verify the updated data types
df[['Year', 'mkt_fare', 'carpax', 'caravgfare']].dtypes
```

The data types of the four columns were successfully converted. `Year` and `carpax` are now stored as integers, while `mkt_fare` and `caravgfare` are stored as floating-point numbers. These conversions enable accurate numerical calculations and statistical analysis of airfare and passenger data.

```python
# Check dataset dimensions, missing values, duplicate rows, and data types
print("Dataset shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum()[df.isnull().sum() > 0])
print("\nDuplicate rows:", df.duplicated().sum())
print("\nData types:")
print(df.dtypes)
```

The final data-quality check confirms that the dataset contains 15,481 rows and 23 columns, with no duplicate records. ###  Bivariate Analysis

Bivariate analysis examines the relationship between two variables to identify differences and patterns in airfare across years, airlines, and city pairs.

```python
# Average market fare by airline
airline_fare = (
    df.groupby('airline')['market_fare']
      .agg(['mean', 'median', 'count'])
      .sort_values('mean', ascending=False)
      .round(2)
)

# Average market fare by quarter
quarter_fare = (
    df.groupby('quarter')['market_fare']
      .agg(['mean', 'median', 'count'])
      .round(2)
)

print("Average Market Fare by Airline:")
display(airline_fare)

print("Average Market Fare by Quarter:")
display(quarter_fare)
```

The quarterly analysis shows a gradual decline in average market fares from **$200.28 in Quarter 1** to **$184.86 in Quarter 4**. Quarter 1 has the highest average fare, while Quarter 4 has the lowest. This suggests possible quarterly variation in airfare across the selected markets.

The airline comparison also shows considerable differences in average market fares. Among the airlines listed, SY has the highest average fare ($277.56), but this is based on only 9 records. NW ($252.75) and DL ($245.30) also have relatively high average fares and substantially more records. At the lower end, WN has an average fare of $114.96 across 3,095 records, while AS averages $106.32 across 492 records.

These differences should be interpreted cautiously because airlines have different numbers of records and may serve different markets and periods. The results describe the selected markets in this dataset and do not establish that an airline is universally more or less expensive.

###  Multivariate Analysis

Multivariate analysis examines multiple variables together to understand how airfare varies across time periods and airlines. In this step, we compare average market fares and passenger counts across years, quarters, and airlines using grouped summaries and a pivot table.

```python
# Average fare and passenger count by year, quarter, and airline
multi_analysis = (
    df.groupby(['Year', 'quarter', 'airline'])
      .agg(
          avg_market_fare=('market_fare', 'mean'),
          avg_passenger_count=('passenger_count', 'mean'),
          records=('market_fare', 'size')
      )
      .reset_index()
)

display(multi_analysis.head(10))

# Pivot table: average market fare by year and quarter
fare_pivot = pd.pivot_table(
    df,
    values='market_fare',
    index='Year',
    columns='quarter',
    aggfunc='mean'
).round(2)

display(fare_pivot.head(10))
```

The grouped analysis compares average market fares and passenger counts across years, quarters, and airlines. The pivot table shows that average fares vary across both years and quarters. For example, in 2000, the average market fare declined from $244.28 in Quarter 1 to $221.55 in Quarter 4. In 2004, it decreased from $173.32 in Quarter 1 to $130.64 in Quarter 4.

These results indicate that airfare patterns differ across time periods. The grouped summary also allows airline-level comparisons within each year and quarter. Further investigation is needed to determine whether these patterns are consistent over time and how fares relate to passenger activity.

###  Correlation Analysis

Correlation analysis measures the strength and direction of linear relationships between numerical variables. In this step, we examine the relationships among market fare, airline average fare, passenger count, passenger market share, and fare-related attributes.

```python
# Correlation matrix for selected numerical variables
corr_cols = [
    'market_fare',
    'airline_avg_fare',
    'passenger_count',
    'passenger_market_share',
    'fareinc_min',
    'fareinc_max',
    'fareinc_minpaxsh',
    'fare_inc_maxpaxsh',
    'fare_inc_x3paxsh'
]

correlation_matrix = df[corr_cols].corr().round(2)
correlation_matrix
```

### Interpretation of Correlation Analysis

The correlation analysis reveals several relationships between airfare, passenger numbers, market share, and fare-increase indicators.

- **Market fare and airline average fare (0.96):** A very strong positive correlation indicates that higher market fares are closely associated with higher airline average fares.
- **Market fare and maximum fare increase (0.86):** A strong positive correlation suggests that higher market fares tend to be associated with higher values of `fareinc_max`.
- **Market fare and minimum fare increase (0.71):** A strong positive relationship is also observed with `fareinc_min`.
- **Market fare and passenger count (-0.24):** A weak negative correlation suggests that markets with higher passenger counts tend to have slightly lower fares, although the relationship is not strong.
- **Market fare and passenger market share (-0.01):** The correlation is almost zero, indicating virtually no linear relationship between these two variables.
- **Airline average fare and passenger count (-0.22):** A weak negative relationship is observed between average fares and passenger numbers.
- **Fare-increase passenger-share indicators:** `fareinc_minpaxsh` has a negative correlation with market fare (-0.72), while `fare_inc_x3paxsh` has a positive correlation (0.55).

**Conclusion:** Airfare measures and fare-increase indicators show several strong relationships, while passenger count and passenger market share have weaker relationships with market fare. Correlation indicates association, not causation, and does not account for other factors that may influence fares.
