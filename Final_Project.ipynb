##  Introduction

The airline industry plays an important role in transportation by connecting people and cities. Airfares can vary depending on the airline, travel market, passenger demand, and time period. Analyzing historical airfare data can help us understand these variations and identify trends in airline markets.

This project focuses on analyzing U.S. domestic airfares and airline market performance using the Consumer Airfare Report – Table 5 dataset published by the U.S. Department of Transportation. The dataset provides detailed fare information for selected high and low-fare markets with distances under 750 miles.

Using Python and its data analysis libraries, this project will explore airfare trends, passenger activity, airline market share, and differences between city-pair markets over time. The analysis will include data cleaning, preprocessing, exploratory data analysis, statistical summaries and visualizations to identify meaningful patterns and insights.

## Attribute Details

The dataset contains 29 attributes that describe the reporting period, airline, city-pair markets, passenger traffic, average fares, fare increments, and geographic information. The following table explains the purpose of each attribute.

| Attribute Name | Description |

| car | The airline carrier's code. |

| carpaxshare | The proportion of market passengers carried by the airline. |
| caravgfare | The average fare charged by the specified airline in the market. |

| fareinc_minpaxsh | The passenger share associated with the minimum fare increment. |
| fareinc_max | The maximum fare increment or fare increase associated with the market. |
| fare_inc_maxpaxsh | The passenger share associated with the maximum fare increment. |
| fare_inc_x3paxsh | The passenger share associated with the fare increment represented by this field in the source dataset. Its precise definition should be confirmed against the source documentation. |
| Geocoded_City1 | Geographic coordinates for the first city, represented as a point. |
| Geocoded_City1 (address) | Address information associated with the first city's geographic record, where available. |
| Geocoded_City1 (city) | City name associated with the first city's geographic record. |
| Geocoded_City1 (state) | State information associated with the first city's geographic record, where available. |

| Geocoded_City2 (state) | State information associated with the second city's geographic record, where available. |
| Geocoded_City2 (zip) | ZIP code associated with the second city's geographic record, where available. |
| tbl5pk | A composite key used to identify a record using the year, market identifiers, carrier identifier, and carrier code. |

###  Removing Completely Empty Columns

```python
# Remove columns where all values are missing
df.dropna(axis=1, how='all', inplace=True)

# Check the updated dataset dimensions
df.shape
```

***Info:-*** After removing the completely empty columns, the dataset contains 15,481 rows and 23 columns.

###  Data Type Inspection and Conversion


```python
# Inspect values before converting data types
df[['Year', 'mkt_fare', 'carpax', 'caravgfare']].head(10)
```

```python
# Remove formatting characters and convert columns to numeric types
df['Year'] = pd.to_numeric(df['Year'].str.replace(',', '', regex=False))
df['mkt_fare'] = pd.to_numeric(df['mkt_fare'].str.replace('$', '', regex=False).str.replace(',', '', regex=False))
df['carpax'] = pd.to_numeric(df['carpax'].str.replace(',', '', regex=False))
df['caravgfare'] = pd.to_numeric(df['caravgfare'].str.replace('$', '', regex=False).str.replace(',', '', regex=False))

# Verify the updated data types
df[['Year', 'mkt_fare', 'carpax', 'caravgfare']].dtypes
```

The data types of the four columns were successfully converted. `Year` and `carpax` are now stored as integers, while `mkt_fare` and `caravgfare` are stored as floating-point numbers. These conversions enable accurate numerical calculations and statistical analysis of airfare and passenger data.

### Renaming Columns for Readability

Selected column names are renamed to improve readability and make the dataset easier to work with during exploratory data analysis. Clear and descriptive names help maintain consistency throughout the analysis.

```python
df.rename(columns={
    'mkt_fare': 'market_fare',
    'car': 'airline',
    'carpax': 'passenger_count',
    'carpaxshare': 'passenger_market_share',
    'caravgfare': 'airline_avg_fare'
}, inplace=True)
```

### Text Standardisation and Whitespace Trimming

```python
# Remove leading and trailing spaces from text columns
text_cols = df.select_dtypes(include=['object', 'string']).columns

for col in text_cols:
    df[col] = df[col].str.strip()

# Standardise the missing-value label
geocoded_cols = [
    'Geocoded_City1',
    'Geocoded_City1 (city)',
    'Geocoded_City2',
    'Geocoded_City2 (city)'
]

for col in geocoded_cols:
    df[col] = df[col].replace('', 'Unknown')
```

### Removing Unnecessary column

```python
# Remove the unnecessary 'tbl' column
df.drop(columns=['tbl'], inplace=True)

# Check the updated dataset dimensions
print("Updated dataset shape:", df.shape)
```

***Reason:-*** Source table label; no analytical value for this single-table project.

### Replacing Missing Values With "UNKNOWN"

```python
# Replace missing geocoded values with 'Unknown'
geocoded_cols = [
    'Geocoded_City1',
    'Geocoded_City1 (city)',
    'Geocoded_City2',
    'Geocoded_City2 (city)'
]

df[geocoded_cols] = df[geocoded_cols].fillna('Unknown')
```

```python
# Check missing values after replacement
print(df.isnull().sum())
```

### Numerical Data Validation

```python
# Check for invalid numerical values

print("Invalid years:",
      (~df['Year'].between(1996, 2026)).sum())

print("Invalid quarters:",
      (~df['quarter'].isin([1, 2, 3, 4])).sum())

print("Non-positive market fares:",
      (df['market_fare'] <= 0).sum())

print("Non-positive airline average fares:",
      (df['airline_avg_fare'] <= 0).sum())

print("Negative passenger counts:",
      (df['passenger_count'] < 0).sum())

print("Passenger market share outside 0–1:",
      (~df['passenger_market_share'].between(0, 1)).sum())
```

### Creating Derived Column

A Year_Quarter column combines the year and quarter to make time-based analysis easier.

```python
# Create a Year_Quarter column
df['Year_Quarter'] = (
    df['Year'].astype(str) + '-Q' + df['quarter'].astype(str)
)

# Display the result
df[['Year', 'quarter', 'Year_Quarter']].head()
```

### Final Data Quality Check

```python
# Check dataset shape, missing values, and duplicates
print("Dataset shape:", df.shape)
print("\nMissing values:\n", df.isnull().sum()[df.isnull().sum() > 0])
print("\nDuplicate rows:", df.duplicated().sum())
```

## Section II: Long-Term and Seasonal Fare Trends

This section explores changes in average market fares over time and compares quarterly patterns to identify long-term trends and potential seasonal variations.

### (i) Yearly Trends in Average Market Fares

- **Visualization Type:** Line Chart (Seaborn)
- **Purpose:** To examine long-term changes in average market fares from 1996 to 2026.
- **Insight:** Average market fares fluctuated considerably over the years, reaching their highest levels around 2013–2015 before declining. Fares increased again from 2021 onwards, showing changes in airfare patterns over time.

```python
yearly_fare = df.groupby('Year')['market_fare'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(data=yearly_fare, x='Year', y='market_fare', marker='o')

plt.title('Yearly Trends in Average Market Fares')
plt.xlabel('Year')
plt.ylabel('Average Market Fare ($)')
plt.tight_layout()
plt.show()
```

### (ii) Average Market Fares by Quarter

- **Visualization Type:** Bar Chart (Seaborn)
- **Purpose:** To compare average fares across Q1, Q2, Q3, and Q4.
- **Insight:** Q1 has the highest average market fare, followed by Q2 and Q3. Q4 has the lowest average fare. This suggests a gradual decline in average fares across the quarters.

```python
quarterly_fare = df.groupby('quarter')['market_fare'].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=quarterly_fare, x='quarter', y='market_fare')

plt.title('Average Market Fares by Quarter')
plt.xlabel('Quarter')
plt.ylabel('Average Market Fare ($)')
plt.xticks(ticks=[0, 1, 2, 3], labels=['Q1', 'Q2', 'Q3', 'Q4'])
plt.tight_layout()
plt.show()
```

### (iii) Quarterly Fare Trends Over Time
- **Visualization Type:** Multi-Line Chart (Seaborn)
- **Purpose:** To compare quarterly fare changes across different years.
- **Insight:** Quarterly fares fluctuate over time, but all quarters show a noticeable increase around 2013–2015, followed by a decline. In recent years, fares have generally increased again, although patterns differ between quarters.

```python
quarterly_trends = (
    df.groupby(['Year', 'quarter'])['market_fare']
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 6))
sns.lineplot(
    data=quarterly_trends,
    x='Year',
    y='market_fare',
    hue='quarter',
    marker='o',
    palette='tab10'
)

plt.title('Quarterly Fare Trends Over Time')
plt.xlabel('Year')
plt.ylabel('Average Market Fare ($)')
plt.legend(title='Quarter')
plt.tight_layout()
plt.show()
```

### (iv) Quarterly Market Fare Trends for Selected Airlines

- **Visualization Type:** Multi-Line Chart (Seaborn)
- **Purpose:** To compare quarterly market fare patterns across selected airlines and identify potential seasonal differences.
- **Insight:** The chart shows differences in average market fares across the five selected airlines and quarters. DL and US have relatively high average fares, while WN consistently has much lower average fares. AA and UA show a gradual decline from Q2 to Q4, whereas DL increases slightly in Q4 after declining through Q3. These are overall quarterly averages across the available records, so they suggest possible seasonal patterns but do not establish that seasonality causes the differences.

```python
top_airlines = (
    df.groupby('airline')
    .size()
    .sort_values(ascending=False)
    .head(5)
    .index
)

quarterly_airline_fares = (
    df[df['airline'].isin(top_airlines)]
    .groupby(['quarter', 'airline'])['market_fare']
    .mean()
    .reset_index()
)

plt.figure(figsize=(10, 6))

sns.lineplot(
    data=quarterly_airline_fares,
    x='quarter',
    y='market_fare',
    hue='airline',
    marker='o'
)

plt.title('Quarterly Market Fare Trends for Selected Airlines')
plt.xlabel('Quarter')
plt.ylabel('Average Market Fare ($)')
plt.xticks([1, 2, 3, 4], ['Q1', 'Q2', 'Q3', 'Q4'])
plt.legend(title='Airline')
plt.tight_layout()
plt.show()
```

This shows that WN accounts for a much larger passenger volume than the other airlines among the selected records

## Section IV: City-Pair Fare Analysis

This section examines average market fares across different city pairs to identify routes with relatively high and low fares and explore differences in pricing across domestic markets.

**City-Pair Sample Size Check**

To improve the reliability of fare comparisons, city pairs are checked for the number of available records. Only city pairs with at least 20 records are included when identifying the highest and lowest average fares.

```python
city_pair_counts = (
    df.groupby(['city1', 'city2'])
    .agg(
        records=('market_fare', 'size'),
        avg_market_fare=('market_fare', 'mean')
    )
    .reset_index()
)

reliable_city_pairs = city_pair_counts[
    city_pair_counts['records'] >= 20
]
```

### (i) Top 10 City Pairs with the Highest Average Market Fares

- **Visualization Type:** Horizontal Bar Chart (Seaborn)
- **Purpose:** To identify the 10 city pairs with the highest average market fares.
- **Insight:** Knoxville, TN–Washington, DC has the highest average market fare among the selected city pairs with at least 20 records, at $326.75. Chicago, IL–Fayetteville, AR follows at $319.23. These results provide a more reliable comparison than including routes with very few records, although the threshold does not guarantee that every route is equally representative.

```python
top_10_reliable = (
    reliable_city_pairs
    .sort_values('avg_market_fare', ascending=False)
    .head(10)
    .copy()
)

top_10_reliable['City Pair'] = (
    top_10_reliable['city1'] + ' - ' + top_10_reliable['city2']
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_10_reliable,
    x='avg_market_fare',
    y='City Pair'
)

plt.title('Top 10 City Pairs with the Highest Average Fares (At Least 20 Records)')
plt.xlabel('Average Market Fare ($)')
plt.ylabel('City Pair')
plt.tight_layout()
plt.show()
```

### (ii) Top 10 City Pairs with the Lowest Average Market Fares

- **Visualization Type:** Vertical Bar Chart (Seaborn)
- **Purpose:** To identify the 10 city pairs with the lowest average market fares.
- **Insight:** Louisville, KY–St. Louis, MO has the lowest average market fare among the selected city pairs with at least 20 records, at $73.45. Reno, NV–San Diego, CA follows at $73.87. The remaining routes have average fares ranging from approximately $76 to $82. These results highlight relatively low-fare routes while reducing the influence of city pairs with very few observations.

```python
bottom_10_reliable = (
    reliable_city_pairs
    .sort_values('avg_market_fare', ascending=True)
    .head(10)
    .copy()
)

bottom_10_reliable['City Pair'] = (
    bottom_10_reliable['city1'] + ' - ' + bottom_10_reliable['city2']
)

plt.figure(figsize=(14, 7))

sns.barplot(
    data=bottom_10_reliable,
    x='City Pair',
    y='avg_market_fare'
)

plt.title('Top 10 City Pairs with the Lowest Average Fares (At Least 20 Records)')
plt.xlabel('City Pair')
plt.ylabel('Average Market Fare ($)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### (iii) Distribution of Market Fares Across Selected City Pairs

- **Visualization Type:** Box Plot (Seaborn)
- **Purpose:** To compare fare distributions across selected city pairs and identify variation and potential outliers.
- **Insight:** The fare distributions vary across the selected city pairs. Cincinnati, OH–New York City, NY shows a relatively wide fare range, while Louisville, KY–New York City, NY has a narrower distribution. Chicago, IL–Cincinnati, OH and Cleveland, OH–Minneapolis/St. Paul, MN contain several potential outliers. These differences suggest that fare variability depends on the route.

```python
top_city_pairs = (
    df.groupby(['city1', 'city2'])
    .size()
    .sort_values(ascending=False)
    .head(5)
    .reset_index(name='records')
)

df['City Pair'] = df['city1'] + ' - ' + df['city2']

selected_pairs = [
    f"{row['city1']} - {row['city2']}"
    for _, row in top_city_pairs.iterrows()
]

plot_data = df[df['City Pair'].isin(selected_pairs)]

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=plot_data,
    x='City Pair',
    y='market_fare'
)

plt.title('Distribution of Market Fares Across Selected City Pairs')
plt.xlabel('City Pair')
plt.ylabel('Market Fare ($)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### (iv) Relationship Between Number of Records and Average Market Fare by City Pair

- **Visualization Type:** Scatter Plot (Seaborn)
- **Purpose:** To examine whether city pairs with more records tend to have different average market fares.
- **Insight:** Most city pairs have fewer than 25 records, with average fares varying considerably across routes. City pairs with higher record counts also show differences in average fares, indicating that the number of records alone does not explain fare variation. Other factors related to the routes may influence average fares.

```python
city_pair_summary = (
    df.groupby(['city1', 'city2'])
    .agg(
        avg_market_fare=('market_fare', 'mean'),
        records=('market_fare', 'size')
    )
    .reset_index()
)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=city_pair_summary,
    x='records',
    y='avg_market_fare',
    alpha=0.6
)

plt.title('Number of Records vs Average Market Fare by City Pair')
plt.xlabel('Number of Records')
plt.ylabel('Average Market Fare ($)')
plt.tight_layout()
plt.show()
```

### (v) Average Market Fare Trends Over Time for Selected City Pairs

- **Visualization Type:** Line Chart (Seaborn)
- **Purpose:** To examine how average market fares change over time for selected city pairs.
- **Insight:** Average market fares fluctuated across all five city pairs over time. Several routes experienced a decline around 2019–2021, followed by an increase in subsequent years. Cleveland, OH–Minneapolis/St. Paul, MN shows a particularly high fare in 2026, while Chicago, IL–Cincinnati, OH shows a sharp decline in its final available observations. These trends should be interpreted cautiously because the number of records varies by route and 2026 contains only first-quarter data.

```python
top_city_pairs = (
    df.groupby(['city1', 'city2'])
    .size()
    .sort_values(ascending=False)
    .head(5)
    .reset_index()
)

df['City Pair'] = df['city1'] + ' - ' + df['city2']

selected_pairs = [
    f"{row['city1']} - {row['city2']}"
    for _, row in top_city_pairs.iterrows()
]

trend_data = df[df['City Pair'].isin(selected_pairs)]

yearly_fares = (
    trend_data.groupby(['Year', 'City Pair'])['market_fare']
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=yearly_fares,
    x='Year',
    y='market_fare',
    hue='City Pair',
    marker='o'
)

plt.title('Average Market Fare Trends Over Time for Selected City Pairs')
plt.xlabel('Year')
plt.ylabel('Average Market Fare ($)')
plt.legend(title='City Pair', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## Section V: Correlation and Relationship Analysis

This section examines relationships between numerical variables in the airfare dataset. Correlation analysis helps identify variables that move together and understand how market fares are associated with other numerical factors.

### (i) Correlation Heatmap of Numerical Variables

- **Visualization Type:** Heatmap (Seaborn)
- **Purpose:** To examine correlations between numerical variables and understand how market fares are associated with other factors.
- **Insight:** The heatmap showed a strong positive correlation between market fare and airline average fare (0.96), and between market fare and `fareinc_max` (0.86). Market fare had a negative correlation with `fareinc_minpaxsh` (-0.72), while passenger count showed a weak negative correlation (-0.24). Passenger market share had almost no linear correlation with market fare (-0.01). These results indicate associations, not causation.

```python
# Select relevant numerical variables for correlation analysis
correlation_cols = [
    'market_fare',
    'airline_avg_fare',
    'passenger_count',
    'passenger_market_share',
    'fareinc_min',
    'fareinc_max',
    'fareinc_minpaxsh',
    'fare_inc_maxpaxsh',
    'fare_inc_x3paxsh'
]

# Calculate the correlation matrix
correlation_matrix = df[correlation_cols].corr()

# Plot the correlation heatmap
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Variables')
plt.tight_layout()
plt.show()
```

### (ii) Relationship Between Market Fare and Airline Average Fare

- **Visualization Type:** Regression Plot (Seaborn)
- **Purpose:** To examine the relationship between market fare and airline average fare.
- **Insight:** The regression plot shows a strong positive relationship between airline average fare and market fare. Most observations follow the upward trend line, which supports the correlation coefficient of 0.96. However, some points deviate from the trend, indicating that the two fare measures do not match perfectly for every record. This relationship represents an association and does not establish causation.

```python
plt.figure(figsize=(9, 6))

sns.regplot(
    data=df,
    x='airline_avg_fare',
    y='market_fare',
    scatter_kws={'alpha': 0.4},
    line_kws={'color': 'red'}
)

plt.title('Market Fare vs Airline Average Fare')
plt.xlabel('Airline Average Fare ($)')
plt.ylabel('Market Fare ($)')
plt.tight_layout()
plt.show()
```

### (iii) Density of Observations Between Passenger Count and Market Fare

- **Visualization Type:** Hexbin Plot (Matplotlib)
- **Purpose:** To examine the density of observations across passenger counts and market fares.
- **Insight:** The hexbin plot shows that most observations are concentrated at relatively low passenger counts. The highest densities appear around market fares of $230–$280 and $75–$100. A small number of records have very high passenger counts, extending beyond one million. This highlights the variation in passenger volumes and fare levels across the selected markets.

```python
plt.figure(figsize=(10, 6))

plt.hexbin(
    df['passenger_count'],
    df['market_fare'],
    gridsize=35,
    cmap='YlOrRd',
    mincnt=1
)

plt.colorbar(label='Number of Records')
plt.title('Density of Passenger Count vs Market Fare')
plt.xlabel('Passenger Count')
plt.ylabel('Market Fare ($)')
plt.tight_layout()
plt.show()
```

### (iv) Relationship Between Market Fare and Maximum Fare-Increase Threshold

- **Visualization Type:** Joint Plot (Seaborn)
- **Purpose:** To examine the relationship between market fare and the maximum fare-increase threshold.
- **Insight:** The joint plot shows a positive association between the maximum fare-increase threshold and market fare. Higher values of `fareinc_max` are generally associated with higher market fares, although the observations show some variation. The marginal distributions also reveal concentrations of records at lower threshold values and around market fares of $230–$280. This relationship indicates an association, not causation.

```python
sns.jointplot(
    data=df,
    x='fareinc_max',
    y='market_fare',
    kind='scatter',
    height=7,
    joint_kws={'alpha': 0.4}
)

plt.suptitle(
    'Market Fare vs Maximum Fare-Increase Threshold',
    y=1.02
)

plt.show()
```
